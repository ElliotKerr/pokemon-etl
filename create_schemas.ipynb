```python
if OVERWRITE_STAGING_SCHEMA is False:
    print(f'Dropping the Staging schema and the data from the lookups tables to prevent version issues.')

    for table_name in ['evo_chain_updated', 'pokedex_updated']:
        if spark.catalog.tableExists(f'{LOOKUPS_DATABASE_PREFIX}.{table_name}'):
            spark.sql(f'TRUNCATE TABLE {LOOKUPS_DATABASE_PREFIX}.{table_name}')

    spark.sql(f'DROP SCHEMA IF EXISTS {STAGING_DATABASE_PREFIX} CASCADE')
    spark.sql(f'CREATE SCHEMA {STAGING_DATABASE_PREFIX}')
              
    print(f'{STAGING_DATABASE_PREFIX} has been dropped and re-created')
else:
    print(f'{STAGING_DATABASE_PREFIX} has been left as is - this means new data will be appended to existing tables.')
```
